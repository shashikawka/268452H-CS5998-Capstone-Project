```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/MyDrive/CAPSTONE_PROJECT/MetroPT3.csv')

print("Shape:", df.shape)
display(df.head())

df.info()
```

```python
#Missing values
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

display(df.describe())
```

```python
print(df.columns.tolist())
```

```python
df['timestamp'] = pd.to_datetime(df['timestamp'])

print("start:", df['timestamp'].min())
print("end:", df['timestamp'].max())

df[['timestamp']].head()
```

```python
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])
```

```python
print("chronological:", df['timestamp'].is_monotonic_increasing)

time_diff = df['timestamp'].diff().dt.total_seconds()

print(time_diff.describe())
print(time_diff.value_counts().head())
```

```python
continuous_sensors = [
    'TP2',
    'TP3',
    'H1',
    'DV_pressure',
    'Reservoirs',
    'Oil_temperature',
    'Motor_current'
]

display(df[continuous_sensors].describe().T)
```

```python
other_sensors = [
    col for col in df.columns
    if col not in continuous_sensors + ['timestamp']
]

for col in other_sensors:
    print(col, ":", df[col].nunique(), "unique values")
    print(df[col].value_counts().head())
    print()
```

```python
normal_period = df[
    (df['timestamp'] >= '2020-02-10') &
    (df['timestamp'] < '2020-02-11')
]

normal_period[continuous_sensors].plot(
    subplots=True,
    figsize=(14, 14),
    sharex=True
)

plt.suptitle('Sensor Behaviour During a Selected Operating Period')
plt.tight_layout()
plt.savefig(
    '/content/drive/MyDrive/CAPSTONE_PROJECT/01_eda/sensor_behaviour_normal_period.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()
```

```python
# Define the time period to inspect
start_time = pd.to_datetime('2020-04-17 00:00:00')
end_time   = pd.to_datetime('2020-04-19 23:59:59')

# Filter the dataset
failure_view = df[
    (df['timestamp'] >= start_time) &
    (df['timestamp'] <= end_time)
]

print("Number of observations:", len(failure_view))
print("Start:", failure_view['timestamp'].min())
print("End:", failure_view['timestamp'].max())

failure_view.head()
```

```python
correlation = df[continuous_sensors].corr()

display(correlation)
```

```python
fig, ax = plt.subplots(figsize=(9, 7))

im = ax.imshow(correlation)

ax.set_xticks(range(len(continuous_sensors)))
ax.set_yticks(range(len(continuous_sensors)))

ax.set_xticklabels(continuous_sensors, rotation=45, ha='right')
ax.set_yticklabels(continuous_sensors)

fig.colorbar(im)

plt.title('Correlation Between Continuous Sensor Variables')
plt.tight_layout()
plt.savefig(
    '/content/drive/MyDrive/CAPSTONE_PROJECT/01_eda/Correlation Between Continuous Sensor Variables.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()
```

```python
# Documented failure intervals

failure_intervals = pd.DataFrame({
    'failure': [
        'Failure 1',
        'Failure 2',
        'Failure 3',
        'Failure 4'
    ],
    'start': pd.to_datetime([
        '2020-04-18 00:00:00',
        '2020-05-29 23:30:00',
        '2020-06-05 10:00:00',
        '2020-07-15 14:30:00'
    ]),
    'end': pd.to_datetime([
        '2020-04-18 23:59:00',
        '2020-05-30 06:00:00',
        '2020-06-07 14:30:00',
        '2020-07-15 19:00:00'
    ]),
    'type': [
        'Air leak',
        'Air leak',
        'Air leak',
        'Air leak'
    ]
})

display(failure_intervals)
```

```python
for _, failure in failure_intervals.iterrows():

    observations = df[
        (df['timestamp'] >= failure['start']) &
        (df['timestamp'] <= failure['end'])
    ]

    print(
        failure['failure'],
        "|",
        failure['type'],
        "| observations:",
        len(observations)
    )
```

```python
df['failure_period'] = 0

for _, failure in failure_intervals.iterrows():

    mask = (
        (df['timestamp'] >= failure['start']) &
        (df['timestamp'] <= failure['end'])
    )

    df.loc[mask, 'failure_period'] = 1

print(df['failure_period'].value_counts())

print("\nPercentage inside documented failure periods:")
print(df['failure_period'].mean() * 100)
```

```python
failure_start = pd.to_datetime('2020-04-18 00:00:00')
failure_end   = pd.to_datetime('2020-04-18 23:59:00')

view_start = failure_start - pd.Timedelta(days=1)
view_end   = failure_end + pd.Timedelta(days=1)

failure_view = df[
    (df['timestamp'] >= view_start) &
    (df['timestamp'] <= view_end)
].copy()

print("Observations:", len(failure_view))
print("View start:", failure_view['timestamp'].min())
print("View end:", failure_view['timestamp'].max())
```

```python
sensors_to_plot = [
    'TP2',
    'TP3',
    'Oil_temperature',
    'Motor_current'
]

fig, axes = plt.subplots(
    len(sensors_to_plot),
    1,
    figsize=(15, 10),
    sharex=True
)

for ax, sensor in zip(axes, sensors_to_plot):

    ax.plot(
        failure_view['timestamp'],
        failure_view[sensor]
    )

   #highlight failure period
    ax.axvspan(
        failure_start,
        failure_end,
        alpha=0.2,
        label='Documented failure period'
    )

    ax.set_ylabel(sensor)
    ax.legend(loc='upper right')

axes[-1].set_xlabel('Time')

plt.suptitle(
    'Sensor Behaviour Before, During and After Documented Failure 1'
)

plt.tight_layout()
plt.savefig(
    '/content/drive/MyDrive/CAPSTONE_PROJECT/01_eda/sensor_behavior_before_during_after.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()
```
