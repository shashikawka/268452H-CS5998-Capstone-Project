```python
# Documented failure intervals

failure_intervals = pd.DataFrame({
    'failure': [
        'Failure 1',
        'Failure 2',
        'Failure 3',
        'Failure 4'
    ],
    'start': pd.to_datetime([
        '2020-04-18 00:00:00',
        '2020-05-29 23:30:00',
        '2020-06-05 10:00:00',
        '2020-07-15 14:30:00'
    ]),
    'end': pd.to_datetime([
        '2020-04-18 23:59:00',
        '2020-05-30 06:00:00',
        '2020-06-07 14:30:00',
        '2020-07-15 19:00:00'
    ]),
    'type': [
        'Air leak',
        'Air leak',
        'Air leak',
        'Air leak'
    ]
})

display(failure_intervals)
```

```python
##Verify the failure periods exist in the dataset

for _, failure in failure_intervals.iterrows():

    observations = df[
        (df['timestamp'] >= failure['start']) &
        (df['timestamp'] <= failure['end'])
    ]

    print(
        failure['failure'],
        "|",
        failure['type'],
        "| observations:",
        len(observations)
    )
```

OBSERVATION : Total observations during the failure periods : 19954

```python
#introduce failure period column

df['failure_period'] = 0

for _, failure in failure_intervals.iterrows():

    mask = (
        (df['timestamp'] >= failure['start']) &
        (df['timestamp'] <= failure['end'])
    )

    df.loc[mask, 'failure_period'] = 1

print(df['failure_period'].value_counts())

print("\nPercentage inside documented failure periods:")
print(df['failure_period'].mean() * 100)
```

```python
df.head()

#0 → timestamp is outside documented failure intervals

#1 → timestamp falls inside a documented failure interval
```

#Vizualize Documented failure period 1

```python
failure_start = pd.to_datetime('2020-04-18 00:00:00')
failure_end   = pd.to_datetime('2020-04-18 23:59:00')

#add 1 day before and after the failure period
# observation between April 17 and April 19.
# to visually compare sensor behaviour before, during and after the documented failure.
view_start = failure_start - pd.Timedelta(days=1)
view_end   = failure_end + pd.Timedelta(days=1)

failure_view = df[
    (df['timestamp'] >= view_start) &
    (df['timestamp'] <= view_end)
].copy()

print("Observations:", len(failure_view))
print("View start:", failure_view['timestamp'].min())
print("View end:", failure_view['timestamp'].max())
```

```python
sensors_to_plot = [
    'TP2',
    'TP3',
    'H1',
    'DV_pressure',
    'Reservoirs',
    'Oil_temperature',
    'Motor_current'
]

fig, axes = plt.subplots(
    len(sensors_to_plot),
    1,
    figsize=(15, 10),
    sharex=True
)

for ax, sensor in zip(axes, sensors_to_plot):

    ax.plot(
        failure_view['timestamp'],
        failure_view[sensor]
    )

   #highlight failure period
    ax.axvspan(
        failure_start,
        failure_end,
        alpha=0.2,
        label='Documented failure period'
    )

    ax.set_ylabel(sensor)
    ax.legend(loc='upper right')

axes[-1].set_xlabel('Time')

plt.suptitle(
    'Sensor Behaviour Before, During and After Documented Failure 1'
)

plt.tight_layout()
plt.savefig(
    '/content/drive/MyDrive/CAPSTONE_PROJECT/01_eda/sensor_behavior_before_during_after_1st_documented_failure.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()
```
